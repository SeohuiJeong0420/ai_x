```python
def get_pricelen(row):
    prices=row['분양가격(제곱미터)']
    if pd.notnull(prices) and prices.strip()!=' ':
        return len(prices)
    else:
        return 0
    
get_pricelen(df_first.loc[0])
```

```python
#  ★  자리수열 추가
df_first = df_first.copy()
df_first['자리수']= df_first.apply(get_pricelen, axis=1)
df_first
```

```python
def commadrop(row):
    row = row.copy()
    price = row['분양가격(제곱미터)']
    if pd.isna(price):
        row['분양가격(제곱미터)'] == np.nan
    elif (price.strip()==' ')|(price.find(',')):
        row['분양가격(제곱미터)'] == np.nan
    else:
        row['분양가격(제곱미터)'] == price
        
    return row
```

```python
df_first = df_first.apply(commadrop, axis=1)
df_first
```
