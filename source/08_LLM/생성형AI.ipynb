```python
#1. 원하는 모델을 이용하여 나라를 입력하면 해당 나라의 가장 유명한 음식을 추천하는 렝체인을 구현(food_chain)
from dotenv import load_dotenv
import os
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
load_dotenv()
llm = ChatOpenAI(model="gpt-4.1-nano")
prompt_template1 = PromptTemplate(
    template="What is the most famous dish of {country}?Return the name of the dish only",
    input_variables=["country"]
)
output_parser = StrOutputParser()
food_chain = prompt_template1 | llm | output_parser
food_result = food_chain.invoke({"country":"Korea"})
print(food_result)
```

```python
# 2. 음식을 입력하면 레시피를 생성하는 렝체인을 구현(recipe_chain)
prompt_template2 = PromptTemplate(
    template="""What is the recipe of {food}? 
    -Step1. [instruction]   
    -Step2. 
    -Step3. 
    -Step4.
    -Step5. 
    Just return list this {food} recipe in order and please ensure each step starts on a new line and has a blank line between steps.""",
    input_variables=["food"]
)
output_parser = StrOutputParser()
recipe_chain = prompt_template2 | llm | output_parser
recipe_result = recipe_chain.invoke({"food":"Kimchi"})
print(recipe_result)
```

```python
# 3. 위의 두 체인을 연결하여, 나라이름을 입력받아, 해당 나라의 가장 유명한 음식의 레시피를 생성하는 렝체인 프로그램을 구현하시오.
result = recipe_chain.invoke({"food":food_result})
print(food_result)
print(result)
```
